```python
import numpy as np
import matplotlib.pyplot as plt

def simulate_gravitational_lensing(grid_size=400, einstein_radius=1.5):
    """
    Simulates gravitational lensing and Einstein ring formation
    using the point-mass gravitational lens bending equation.
    """
    # Create 2D observer sky coordinate grid (theta_x, theta_y)
    x = np.linspace(-4, 4, grid_size)
    y = np.linspace(-4, 4, grid_size)
    X, Y = np.meshgrid(x, y)

    # Angular distance from lens center (theta)
    theta = np.sqrt(X**2 + Y**2)
    theta = np.maximum(theta, 1e-5) # Prevent division by zero at center

    # Gravitational deflection angle mapping (General Relativity lens equation)
    # Alpha = (4 * G * M) / (c^2 * b) -> scaled by Einstein radius theta_E^2 / theta
    alpha = (einstein_radius**2) / theta

    # Source plane position coordinates (beta = theta - alpha)
    # Vectorized deflection mapping
    X_source = X - alpha * (X / theta)
    Y_source = Y - alpha * (Y / theta)

    # Define a distant background source structure (e.g., a luminous spiral galaxy or dual spots)
    source_intensity = np.exp(-((X_source - 0.3)**2 + (Y_source - 0.3)**2) / 0.15) + \
                       0.7 * np.exp(-((X_source + 0.4)**2 + (Y_source + 0.2)**2) / 0.1)

    # Magnification matrix calculation (determinant of Jacobian)
    # Shows where gravitational amplification occurs (critical curves / caustics)
    der_xx = 1.0 - (einstein_radius**2 / theta**3) * (Y**2 - X**2)
    der_yy = 1.0 - (einstein_radius**2 / theta**3) * (X**2 - Y**2)
    der_xy = - (einstein_radius**2 / theta**3) * (2 * X * Y)
    magnification = np.abs(1.0 / (der_xx * der_yy - der_xy**2))
    magnification = np.clip(magnification, 0, 20) # Cap extreme singularities

    # Lensed image brightness map
    lensed_image = source_intensity * (0.4 + 0.1 * magnification)

    # Plotting
    fig, ax = plt.subplots(figsize=(8, 8), dpi=130)
    im = ax.imshow(lensed_image, extent=[-4, 4, -4, 4], origin='lower', cmap='inferno')

    # Draw central lensing mass / Einstein ring reference circle
    circle = plt.Circle((0, 0), einstein_radius, color='cyan', fill=False, linestyle='--', linewidth=1.5, alpha=0.7, label='Einstein Ring ($\theta_E$)')
    ax.add_patch(circle)

    ax.set_title("Relativistic Gravitational Lensing & Einstein Ring Simulation\n(General Relativity Photon Deflection)", fontsize=11, fontweight='bold')
    ax.set_xlabel("Angular Position $\\theta_x$ (arcsec)")
    ax.set_ylabel("Angular Position $\\theta_y$ (arcsec)")
    ax.legend(loc='upper right')
    plt.tight_layout()
    plt.show()

# Run Simulation
simulate_gravitational_lensing()
```
